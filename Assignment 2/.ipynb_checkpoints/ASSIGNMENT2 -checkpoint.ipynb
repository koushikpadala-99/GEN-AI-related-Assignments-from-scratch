```python
#1 student report card
students={"Rahul":91,
          "Ayesha":84,
          "Kiran":67,
          "Ali":45}
pc,fc=0,0
for key,value in students.items():
    print(key,value)
    if value>=40:
        pc+=1
        print("pass")
    else:
        fc+=1
        print("fail")
print("no.of students passed:",pc)
print("no.of students failed:",fc)
```

```python
#2 grocery store inventory
inventory={"Rice":20,
           "Sugar":15,
           "Milk":8,
           "Eggs":30}
item=input("Enter the item to buy:")
if item in inventory:
    inventory[item]-=1
else:
    print("Item not available")
for key,value in inventory.items():
    print(key,value)
```

```python
#3 login system
users={"admin":"python123",
       "student":"welcome",
       "guest":"guest123"}
i=3

while(i>0):
    username=input("Enter the username:")
    pwd=input("Enter the password:")
    if username in users and users[username]==pwd:
        
        print("Login successful")
        break
    i-=1
print("Account locked")

        
    
    
```

```python
#6 daily expense tracker
Tracker={}
count=0
while True:
    expense_name=input("")
    if expense_name=="stop":
       
        for key,value in Tracker.items():
            print(key,value)
        print(count)
        break
    else:
        amount=int(input())
        Tracker[expense_name]=amount
        count+=1
```

```python
#15 shopping bill generator
items={"Milk":50,
       "Bread":40,
       "Eggs":70,
       "Rice":500}
total=0
while True:
    item=input("enter the item name:")
    if item.lower()=="done":
        break
    elif item in items:
        total+=items[item]
    else:
        print("item not available")
    
print(total)
```

```python
#20 space mission log
systems={
    "Engine":"OK",
    "Fuel":"OK",
    "Navigation":"OK",
    "Communication":"OK"}
flag=False
for key,value in systems.items():
    if value=="FAIL":
        flag=True
        print(key)
        file=open('mission_log.txt','w')
        file.write(key+"\n")
        file.close()
        print("Mission Aborted")
        break
if flag==False:
    file=open('mission_log.txt','w')
    content="Launch Successful"
    file.write(content)
    file.close()
     
```
