```python
password=input("Enter your password")
Upperchar=False
Lowerchar=False
digit=False
for ch in password:
    if ch.isupper():
        Upperchar=True
    elif ch.islower():
        Lowerchar=True
    elif ch.isdigit():
        digit=True
if Upperchar and Lowerchar and digit:
    print("Strong Password")
else: print("Weak Password")
```

```python
chat="Python is fun Python is powerful"
count=0;
list=chat.split()
print(list)
#enumerate tracks the both index and value in the list 
for i,ch in enumerate(list):
    if ch.lower()=="python":
        count+=1
    #only through indexing we can do changes in the list
    if ch.lower()=="powerful":
        list[i]="Amazing"
print(f"Python is repeated: {count} times")
chat=" ".join(list)
print(chat)
```

```python
def withdrawal(balance,amount):
    if balance<amount:
        return "Insufficient balance"
    return balance-amount
withdrawal(10000,100000)
```

```python
list=[]
def bill(price,quantity):
    total=price*quantity
    list.append(total);
    print("----Reciept----")
    print(f"Total amount {total}")
    print("----End---")
    return
bill(20,12.22)
bill(12,22)
print(list)
        
```
