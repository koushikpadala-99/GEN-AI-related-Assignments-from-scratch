```python
team_list=["member 1","member 2","member 3","member 4"]
with open('file1.txt','a+') as file:
    for i in team_list:
        attend=f'enter present or absent for {i}'
        attendence=input(f'enter present or absent for {i}:')
        file.write("\n"+attend+attendence)
    file.seek(0)
    content=file.read()
    file.close()
print(content)
```
