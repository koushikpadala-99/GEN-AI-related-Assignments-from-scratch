```python
a=np.arange(12)
a
```

```python
#reshape is used to convert the dimension of the array
reshaped_arr=a.reshape(3,4)
reshaped_arr
```

```python
#ndim is used to find the dimension of the array
reshaped_arr.ndim
```

```python
print("Print the first row: ",reshaped_arr[0,:])
```

```python
print("Print the first column: ",reshaped_arr[:,0])
```

```python
print(reshaped_arr[0:2,1:3])
```

```python
print(sum(reshaped_arr))
```

```python
print("sum of the elements in the numpy array: ",np.sum(reshaped_arr))
print("mean of the matrix: ",np.mean(reshaped_arr))
print("median of the matrix: ",np.median(reshaped_arr))
print("variance of the matrix: ",np.var(reshaped_arr))
print("minimum of the array",np.min(reshaped_arr))
print("maximum of the array: ",np.max(reshaped_arr))
print("minimum value index in the array/matrix: ",np.argmin(reshaped_arr))#here it first flattens the array and gives the index
print("max value index in the array: ",np.argmax(reshaped_arr))
print("product of the array: ",np.prod(reshaped_arr))
#also we have many functions like no.any()/np.all() etc.....
```

```python
#this concept is for getting the rows and cols in a matrix in argmax/argmin
#here we use the unravel function to get the row and col of the min/max ele in the matrix
index=np.argmax(reshaped_arr)
row,col=np.unravel_index(index,reshaped_arr.shape)
print(row,col)
```
