```python
# Q-6
import time
python_list=list(range(100000))
np_list=np.arange(100000)

start=time.time()
result=[x*2 for x in python_list]
end=time.time()
print("Time taken : ",end-start)
start=time.time()
np_res=np_list*2
end=time.time()
print("time taken by np list : ",end-start)
# numpy operation is faster
```

```python
# Q-7
py_fruits=[1,2,3,4,5]
np_fruits=np.array([1,2,3,4,5])
# for i in range(len(py_fruits)):
#     py_fruits[i]=py_fruits[i]*2
#     np_fruits=np_fruits*2
py_fruits=[i*2 for i in py_fruits]
np_fruits=np_fruits*2
print(py_fruits)
print(np_fruits)
```

```python
# Q-8
stock=[10,20,30,40]
py_stock=[st*2 for st in stock]
np_stock=np.array(stock)*2
print(py_stock)
print(np_stock)
```

```python
# Q-9
exp=[10,20,30,40]
np_exp=np.array(exp)
print(sum(exp))
print(np.sum(np_exp))
```
